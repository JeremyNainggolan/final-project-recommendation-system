## Baseline Model: Item Popularity-Based Recommender

```python
import pandas as pd
```

```python
# Load data
train_df = pd.read_csv('../data/train.csv')
test_users_df = pd.read_csv('../data/data_target_users_test.csv')
sample_submission_df = pd.read_csv('../data/sample_submission.csv')  # For reference
```

```python
# Compute item popularity (total interactions per item)
item_popularity = train_df['item_id'].value_counts().sort_values(ascending=False)
```

```python
# Function to get top 10 recommendations for a user
def get_popularity_recommendations(user_id, top_items, user_interactions):
    # Get items user has already interacted with
    interacted_items = set(user_interactions.get(user_id, []))
    # Recommend top items not interacted with
    recommendations = [item for item in top_items.index if item not in interacted_items][:10]
    return ' '.join(recommendations)
```

```python
# Prepare user interactions dict from train data
user_interactions = train_df.groupby('user_id')['item_id'].apply(list).to_dict()

# Get top 10 global items
top_10_items = item_popularity.head(10)

top_10_items
```

```python
# Generate recommendations for each test user
recommendations = []
for user_id in test_users_df['user_id']:
    recs = get_popularity_recommendations(user_id, top_10_items, user_interactions)
    recommendations.append({'user_id': user_id, 'item_id': recs})
```

```python
# Create submission DataFrame
submission_df = pd.DataFrame(recommendations)
submission_df.to_csv('../data/baseline_submission.csv', index=False)
```
